# Introducción

En esta primera fase del proyecto CostaBlancaFinder AI comenzamos construyendo el núcleo de inteligencia de datos de la plataforma.

La filosofía del proyecto es "Data First", es decir, construir primero la infraestructura de datos antes de desarrollar aplicaciones visuales complejas.

En esta etapa inicial trabajamos con:

- datasets inmobiliarios

- análisis exploratorio

El objetivo es transformar datos simples en inteligencia de negocio escalable.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../processed_data/rentals_scored.csv")

df
```

# Exploración inicial del dataset

En esta sección analizamos el dataset generado por el Opportunity Scoring Engine.

Variables principales:

- city
- zone
- price_eur
- area_m2
- price_m2
- opportunity_score

Estas variables permitirán posteriormente construir sistemas más avanzados de recomendación y detección de oportunidades.

```python
plt.figure(figsize=(10,5))

plt.bar(df["city"], df["opportunity_score"])

plt.title("Opportunity Score by City")
plt.xlabel("City")
plt.ylabel("Opportunity Score")

plt.show()
```

# Interpretación inicial

Observamos que Villajoyosa obtiene actualmente la puntuación de oportunidad más alta.

Esto se debe principalmente a:

- menor precio por metro cuadrado
- mayor número de habitaciones
- proximidad a zonas turísticas

El sistema comienza así a detectar patrones de valor dentro del mercado inmobiliario local.
